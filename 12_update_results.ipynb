```python
from pathlib import Path
import pandas as pd

# Add the completed 2025–26 season and the current 2026–27 season
new_seasons = {
    "2025-26": "2526",
    "2026-27": "2627",
}

raw_data_folder = Path("data/raw")

for season_name, season_code in new_seasons.items():
    url = f"https://football-data.co.uk/mmz4281/{season_code}/E0.csv"

    season_data = pd.read_csv(url, encoding="latin1")
    season_data["Season"] = season_name

    # Save the original download exactly as received
    file_name = raw_data_folder / f"epl_{season_code}.csv"
    season_data.to_csv(file_name, index=False)

    # Count only matches that already have a final score
    completed_matches = season_data.dropna(subset=["FTHG", "FTAG"])

    print(
        f"{season_name}: downloaded {len(season_data)} rows, "
        f"{len(completed_matches)} completed matches"
    )
```
