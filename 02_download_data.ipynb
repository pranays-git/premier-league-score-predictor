```python
from pathlib import Path
import pandas as pd

# Ten completed Premier League seasons
seasons = {
    "2015-16": "1516",
    "2016-17": "1617",
    "2017-18": "1718",
    "2018-19": "1819",
    "2019-20": "1920",
    "2020-21": "2021",
    "2021-22": "2122",
    "2022-23": "2223",
    "2023-24": "2324",
    "2024-25": "2425",
}

raw_data_folder = Path("data/raw")
all_seasons = []

for season_name, season_code in seasons.items():
    url = f"https://football-data.co.uk/mmz4281/{season_code}/E0.csv"

    # Download one season of Premier League results
    season_data = pd.read_csv(url, encoding="latin1")
    season_data["Season"] = season_name

    # Save the untouched source file
    file_name = raw_data_folder / f"epl_{season_code}.csv"
    season_data.to_csv(file_name, index=False)

    all_seasons.append(season_data)
    print(f"Downloaded {season_name}: {len(season_data)} matches")

# Combine all seasons into one table
matches = pd.concat(all_seasons, ignore_index=True)

print("\nTotal rows downloaded:", len(matches))
matches[["Season", "Date", "HomeTeam", "AwayTeam", "FTHG", "FTAG", "FTR"]].head()
```
