```python
import pandas as pd

# Load the saved, cleaned dataset
matches = pd.read_csv(
    "data/processed/epl_matches_clean.csv",
    parse_dates=["Date"]
)

# Basic scoring patterns
matches["TotalGoals"] = matches["FTHG"] + matches["FTAG"]

home_win_rate = (matches["FTR"] == "H").mean()
draw_rate = (matches["FTR"] == "D").mean()
away_win_rate = (matches["FTR"] == "A").mean()

print(f"Average total goals per match: {matches['TotalGoals'].mean():.2f}")
print(f"Average home goals: {matches['FTHG'].mean():.2f}")
print(f"Average away goals: {matches['FTAG'].mean():.2f}")
print()
print(f"Home-win rate: {home_win_rate:.1%}")
print(f"Draw rate:     {draw_rate:.1%}")
print(f"Away-win rate: {away_win_rate:.1%}")

display(
    matches["FTR"]
    .value_counts()
    .rename(index={"H": "Home win", "D": "Draw", "A": "Away win"})
)
```
